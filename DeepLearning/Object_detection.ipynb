<h3> Object Detection </h3>

Finding an object instance in an image, classify the object and create a bounding box <br><br>

<img src="https://miro.medium.com/max/503/1*xWntyXM0W-SuDMgWMM6mCg.png" align="right">

<h4>How does the output of the network look like?</h4>

In the easiest case, one object per image:

b1-b4: b1,b2: coordinate of one edge of the bounding box, b3:height of the bounding box, b4: width of bounding box <br>
c1-cn: One hot encoded vector of the class <br>
p1: Probability that object is included >br>
<br>
--> Vector of length 5+n <br> 
<br>
[b1,b2,b3,b4,p1,c1,c2, ...cn]

Source: Vasilev, 2019

Object detection describes the task of finding object instances in an image. This means the object instance need to be classified as well as detected, in which region the object is located. Compared to a classification and regression problem, the network output is changed for object detection. At a minimum, there are 5+n outputs per object, where n counts the number of classes. C1 to Cn are the one-hot transformed classification of which object is detected, p describes the probability that the object belongs to class c. B1 and B2 are one coordinate of the bounding box, while B3 and B4 are describing the width and height of the bounding box. Thus, the output for each object has the shape of a 5+n dimensional vector (Vasilev, 2019). 

<h3>Naive Approach</h3>
<list>
<li> Divide image into regions </li>
<li> Pass regions to a normal cnn </li>
<li> Bounding box has the coordinates of the region passed to the network </li>
</list>
<br><br>
Downside: To classify objects correctly with many objects in the image, a lot of regions would need to be passed to the network <br><br>This is very inefficient in terms of computation

Source: Vasilev, 2019

<h3>One stage vs two stage approaches</h3>

<h5>One stage approach</h5>
<list>
    <li>Detection and Classification performed in one step/network</li>
    <li>Example: YOLO </li>
</list>

<h5>Two stage approach</h5>
<list>
    <li>Step 1: Proposing interesting regions </li>
    <li>Step 2: Classification and Bounding Box Regression of the regions </li>
    <li>Example: Fast R-CNN, Faster R-CNN </li>
</list>

In general, two different model architectures exist for object detection, One-stage, and two-stage models. While in a two-stage model, the regional proposal of objects is separated from the classification and location task, one-stage models combine everything in one network. The most commonly known one-stage method is YOLO (You only look once). R-CNN, Fast R-CNN, and Faster R-CNN are two-stage methods (Vasilev, 2019).

R-CNN was the first well-performing object detection method. Via selective search, interesting regions in the image are identified. Often 2000 different regions are sampled per image. Each of the 2000 different images are passed through a CNN for object classification. The bounding box does not need to be calculated as the image part sets the bounding box itself. While this method is easy to implement, it comes with the downside of many predictions per image. Thus, it is relatively slow, 45 seconds per image, and no real-time application is possible, for example, autonomous driving. Thus, R-CNN was improved, and Fast R-CNN was developed. While the overall idea stays the same, the whole image is first passed through and CNN. The convoluted activations are derived, and selective search is applied to the convoluted images. Thus, only one pass through the CNN is needed to be performed. A special ROI pooling is performed to save the area of interest, and fully connected layers perform the classification of the object as well as the bounding box prediction. While Fast R-CNN is faster than R-CNN, it might not be fast enough, so Faster R-CNN comes into play. Compared to Fast R-CNN, instead of using selective search, a regional proposal network is trained for suggesting intelligently areas of interest. The Regional Proposal Network is a separately trained CNN for just predicting interesting regions in images (Vasilev, 2019).

<h3> YOLO </h3>
<list>
         <ul class="square">
          <li>One stage method</li>
          <li>Divide image into S by S grid</li>
          <li>Each grid can have m objects; Anchorbox resembles 1-to-m relationship </li>
          <li>CNN does classification and regression for each grid </li>
          <li>Fastest detection algorithm: less than 1 second per image</li>
        </ul>
<img src="https://miro.medium.com/max/1400/1*JniWRt-ceWLNlkOULjhdpg.png" width=500> <br>
Image source: Ghandi, 2018

Yolo is up to now the fastest method for object detection. As it is built on just one neural network, different assumptions need to be made. First, the image is divided into an S by S grid. In each grid, m objects can be found. On each grid, the CNN is applied with the output of m times the 5+n output vector. The Anchorboxes are in a 1 to m relationship so that each object belongs to one anchor box (which can be spread over multiple grids). However, each grid can have m objects. It is with less than 1 second per image the fasted object detection methods (Vasilev, 2019).

<h3>IoU</h3>
Intersection over Union <br>
<br>
Technique to denoise the prediction <br>
Discard box when IoU larger than threshold (mostly 0.5)

<img src="https://pyimagesearch.com/wp-content/uploads/2016/09/iou_equation.png" width=300 height=300>

Source: Rosenbrock, 2016

For the described methods, the IoU concept is used to denoise the predictions. IoU stands for intersection over union, where for all the derived object predictions, the area of overlap is divided by the area of union. Most of the time, a fixed threshold, for example 0.5, is applied, discarding all the bounding boxes and detection above the threshold.  This deletes multiple detections of the same object (Rosenbrock, 2016, Vasilev, 2019).

```python
import numpy as np
import pandas as pd
from tensorflow.keras import layers, models
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.preprocessing.image  import ImageDataGenerator
from matplotlib import pyplot as plt
import matplotlib
from tqdm import tqdm
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, MinMaxScaler
from sklearn.metrics import accuracy_score, mean_absolute_error
from tensorflow.keras.applications.resnet50 import ResNet50
import os
from skimage import io, transform, draw
import glob
import gc
```

```python
def read_images(data, org_path, dim):
    x=[] # empty list for returing the images
    y=[] # empty list for returning the label
    pos=[] # empty list for returning the position
    classes=os.listdir(org_path) # all subfolders in the org_path
    
    for c in classes: # for each class
        print("Reading class: " + c) # Print reading which class
        path=os.path.join(org_path, c) # creating the subpath for the folder
        
        data_2=data[data.ClassId==int(c)] # get the data from the data frame containing the positions and image names
        
        for i in range(0, len(data_2)): # for all images belonging to that class
            
            datatemp=data_2.iloc[i,:]
                                        # Get the ROI
            temp_x1=datatemp["Roi.X1"] 
            temp_y1=datatemp["Roi.Y1"]
            
            temp_x2=datatemp["Roi.X2"]
            temp_y2=datatemp["Roi.Y2"]
            
            img=datatemp["Filename"]
            
            img_path=os.path.join(path, img) # create the image path
            
            
            image=io.imread(img_path) # read the image; here with scikit image
                                    # Rescale the ROI and images
            if image is not None:
                # extract with and height
                w=image.shape[1]
                w=int(w)
                h=image.shape[0]
                h=int(h)
                
                # Rescaling can just work, if not on boundary --> > 0 
                # thus add one or small number to the bounding box coordinates
                
                temp_x1=temp_x1+1
                temp_y1=temp_y1+1
                
                temp_x2=temp_x2+1
                temp_y2=temp_y2+1
                
                # rescale it to the output dimension dim
                x1=(int(temp_x1)/w)*dim[1]
                y1=(int(temp_y1)/h)*dim[0]


                x2=(int(temp_x2)/w)*dim[1]
                y2=(int(temp_y2)/h)*dim[0]
            
            
                
                scaled=transform.resize(image, dim) # scale the image
                x.append(scaled) # append image 
                y.append(c) # append the class 
                pos.append([x1,y1,x2,y2]) # append the positions

    return np.array(x, dtype=np.float16), y, pos
```

```python
image_path="./GTSRB/Final_Training/Images"
data_path="./GTSRB/Final_Training/Images"

print(os.listdir("./GTSRB"))
```

```python
df = pd.concat([pd.read_csv(f, sep=";") for f in glob.glob(data_path+'/**/*.csv', recursive=True)], ignore_index = True) # read the meta data
```

```python
print(df.info()) # contains filename, classId and bounding box coordinates
```

```python
# Check class distribution 
print(df.ClassId.value_counts(normalize=True))
```

```python
print(df["Roi.X1"].min())
print(df["Roi.X1"].max())
print(df["Roi.X2"].min())
print(df["Roi.X2"].max())

print(df["Roi.Y1"].min())
print(df["Roi.Y1"].max())
print(df["Roi.Y2"].min())
print(df["Roi.Y2"].max())
```

```python
# read images and scaled bounding boxes
x,y, pos = read_images(data=df,org_path=image_path, dim=(100,100))

y=pd.DataFrame(y)
y=np.reshape(y,(-1,1))

print(y.shape)

# One hot encoding for classification
onh=OneHotEncoder(sparse=False)
onh.fit(y)
y=onh.transform(y)
```

```python
print(y.shape)
```

```python
# Train-test-split

X_train, X_test, y_train, y_test, X_pos_train, X_pos_test = train_test_split(x,y,pos, test_size=0.2, random_state=42)

del x,y,pos
gc.collect() # delete unnecessary data and clear cache

y_train = np.array(y_train)
y_test = np.array(y_test)

# Noramlize the images
X_train=X_train/255.0
X_test=X_test/255.0
```

```python
# The coordinates X1, y1 -> first corner and x2, y2 -> second corner are now mapped between 0 and 1

X_pos_train=pd.DataFrame(X_pos_train)
X_pos_test=pd.DataFrame(X_pos_test)

# X1
X_pos_train_x1=np.array(X_pos_train.iloc[:,0]).reshape(-1, 1)
X_pos_test_x1=np.array(X_pos_test.iloc[:,0]).reshape(-1, 1)

minmax_x1 = MinMaxScaler()
X_pos_train_x1=minmax_x1.fit_transform(X_pos_train_x1)
X_pos_test_x1=minmax_x1.transform(X_pos_test_x1)

# X2
X_pos_train_x2=np.array(X_pos_train.iloc[:,2]).reshape(-1, 1)
X_pos_test_x2=np.array(X_pos_test.iloc[:,2]).reshape(-1, 1)

minmax_x2 = MinMaxScaler()
X_pos_train_x2=minmax_x2.fit_transform(X_pos_train_x2)
X_pos_test_x2=minmax_x2.transform(X_pos_test_x2)




# Y1
X_pos_train_y1=np.array(X_pos_train.iloc[:,1]).reshape(-1, 1)
X_pos_test_y1=np.array(X_pos_test.iloc[:,1]).reshape(-1, 1)

minmax_y1 = MinMaxScaler()
X_pos_train_y1=minmax_y1.fit_transform(X_pos_train_y1)
X_pos_test_y1=minmax_y1.transform(X_pos_test_y1)


# Y2
X_pos_train_y2=np.array(X_pos_train.iloc[:,3]).reshape(-1, 1)
X_pos_test_y2=np.array(X_pos_test.iloc[:,3]).reshape(-1, 1)

minmax_y2 = MinMaxScaler()
X_pos_train_y2=minmax_y2.fit_transform(X_pos_train_y2)
X_pos_test_y2=minmax_y2.transform(X_pos_test_y2)
```

```python
print(y_train.shape)
print(y_train[1])
print(onh.inverse_transform(y_train)[1])
```

```python
############## Your codes goes here ############################
# Inputs are X_train
# Targest are y_train, X_pos_train_x1, X_pos_train_y1, X_pos_train_x2, X_pos_train_y2

def model():
    img_input=layers.Input(shape=(100,100,3))

    _model=layers.Conv2D(32, (5,5), activation="relu") (img_input)
    _model=layers.MaxPooling2D() (_model)

    _model=layers.Conv2D(64,(3,3), activation="relu") (_model)
    _model=layers.MaxPooling2D() (_model)

    _model=layers.Conv2D(128,(3,3), activation="relu") (_model)
    _model=layers.MaxPooling2D() (_model)

    _model=layers.GlobalMaxPooling2D() (_model)
    _model=layers.Dense(256, activation="relu") (_model)
    _model=layers.Dense(128, activation="relu") (_model)

    # define outputs
    class_prediction=layers.Dense(y_train.shape[1], activation="softmax", name="class_prediction") (_model)
    x1_prediction=layers.Dense(1,activation="linear", name="x1_prediction") (_model)
    y1_prediction=layers.Dense(1,activation="linear", name="y1_prediction") (_model)

    x2_prediction=layers.Dense(1,activation="linear", name="x2_prediction") (_model)
    y2_prediction=layers.Dense(1,activation="linear", name="y2_prediction") (_model)

    # optimizer + build model
    opt=Adam(lr=0.001)
    full_model=models.Model(img_input,[class_prediction, x1_prediction, y1_prediction, x2_prediction, y2_prediction])
    full_model.compile(optimizer=opt, loss={'class_prediction': 'categorical_crossentropy','x1_prediction': 'mae', 'x2_prediction': 'mae', 'y1_prediction': 'mae', 'y2_prediction': "mae"}, metrics={'class_prediction': 'accuracy', 'x1_prediction': 'mae'})
    
    print(full_model.summary())
    return full_model
  
prediction_model = model()
prediction_model.fit(X_train, [y_train, X_pos_train_x1, X_pos_train_y1, X_pos_train_x2, X_pos_train_y2], epochs=25, validation_split=0.1)
```

```python
predictions = prediction_model.predict(X_test)
print(len(predictions))
```

```python
prediction_class = predictions[0]
predictions_pos_x1 = predictions[1]
predictions_pos_y1 = predictions[2]
predictions_pos_x2 = predictions[3]
predictions_pos_y2 = predictions[4]
```

```python
from sklearn.metrics import mean_absolute_error
```

```python
print(accuracy_score(onh.inverse_transform(y_test), onh.inverse_transform(prediction_class)))
print(mean_absolute_error(X_pos_test_x1, predictions_pos_x1))
print(mean_absolute_error(X_pos_test_y1, predictions_pos_y1))
print(mean_absolute_error(X_pos_test_x2, predictions_pos_x2))
print(mean_absolute_error(X_pos_test_y2, predictions_pos_y2))
```
