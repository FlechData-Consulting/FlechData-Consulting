<h2>Adversarial example</h2>

Example (image) for which a false prediction is made be a machine learning algorithms with just little pertubations of the original example

The example fools the machine learning with differences that are not visible for the human eye

The statement can be formally be written as:

$$ \underset{x'}{\text{min}}\vert \vert x'-x \vert \vert $$
$$ \textrm{s.t.} \quad f(x')=l' \\ $$
$$ \quad f(x)=l \\ $$
$$ \quad l \neq l' \\ $$
$$ \quad x' \in [0,1] \\ $$

Source: Yuan et al. (2018)

<h2> Adversarial attacks </h2>

An adversarial attack is the useage of adversarial examples on an existing neural network <br><br>
There are different scales on which adversarial attacks are rated: <br>

<h3>Adversarial Falsification</h3>
<list>
    <li> False positiv: Aim to generate Typ 1 errors
        <ul>Generate negative example to be misclassified as a positive one</ul>
        <ul>Images unrecognizable to a human (noise) classified with a high confidence into a certain class </ul>
    </li>
    <li> False negativ: Aim to generate Typ 2 errors

<h3>Pertubation Measures</h3>
$\ell_p$ is the p-normed distance between the original image and the adversarial example

$$ \vert \vert x \vert \vert_p = (\sum_{i=1}^n \vert \vert x_i \vert \vert^P)^\frac{1}{p}$$

<list>
    <li> $\ell_0$ count the number of pixels changed in the adversarial example</li>
    <li> $\ell_2$ measures the euclidean distance between image and example </li>
     <li> $\ell_\infty$ maximum change for all pixels </li>
</list>

Source: Yuan et al. (2018)

In the recent development in Computer Vision and Deep Learning, often, the question arises of how trustworthy and reliable the results of neural networks are. Especially when it comes to security-related aspects of solutions in autonomous driving, medicine, or face recognition solutions should be very mature to guarantee accurate predictions. Research about adversarial attacks against machine learning algorithms has shown that these solutions may not be as mature as necessary. An adversarial is a generated input, in CV, an image that is based on an original image plus some noise created by perturbations. The noise is unrecognizable for the human eye. Thus, the neural network gets confused in the prediction so that it predicts the wrong label of the image. Adversarial attacks have the aim to test and confuse the neural network's decision making by adding as little noise as possible to the image. 

There are several types of adversarial attacks distinguished by different scales:

An attack can be rated on the falsification strategy, whether the networks make Type 1 or Type 2 errors. Type 1 errors are made when negative observations are falsely predicted as positive. In case of an attack, the pure noise which has no structure is categorized to a certain class e.g. a car. Type 2 errors are made when there are positive examples, the adversarial example shows a car with noise unrecognizable to the human eye, and it is wrongly predicted, e.g., as a dog. Moreover, attacks are differentiated in white and black-box attacks, where most of the attacks are black-box attacks as the adversarial has access to the network's structure and parameters.  While non-targeted attacks have the aim that the adversarial example is misclassified, targeted attacks aim for misclassification with a specific target (class). Most of the attacks are individual attacks, so that for each example in the dataset, the noise is individually calculated, while universal attacks create one noise for the whole dataset. Thus, these are often used for white-box attacks. Lastly, attacks can be distinguished on how the perturbations are calculated. Either they are part of the optimization aim, thus called optimized perturbations or the perturbations are part of the constraints and thus need to be "good" enough to fool the network. 

Source: Yuan et al. (2018), Szegedy et al. (2013)

<h3>Fast Gradient Sign Method (FGSM)</h3>

To calculate the pertubations is often a complex and expensive linear search method <br>
FGSM is a technique to increase efficiency by making use of the gradient of the weights and linearity between adversarial noise and image <br>
<br>

$$ x'=x+\eta $$ <br>
$$ \eta = -\epsilon *sign(\nabla_x J_\theta(x,l)) $$

$\epsilon$ is a small number and accounts for the magnitude of the pertubation

Goodfellow, Shlens, and Szegedy came up with a fast method to generate adversarial examples. Their idea is to use the network properties itself to find vulnerable spaces in the network. Therefore, the gradient for a specific input image is calculated with respect to its output. The gradients of the loss function contain information about the importance of each pixel. Therefore a small noise in form of epsilon is added or subtracted from the original image, depending on the sign of the pixel gradient. Adding errors in the direction of the gradients means increasing the loss, thus confusing the algorithm. Especially vulnerable are networks that favor linearity like LSTM or maxout networks and relu based networks. 

Source: Goodfellow, Shlens, Szegedy (2014)

```python
# Import packages
import numpy as np
from tensorflow.keras import layers, metrics, optimizers, models, utils
from tensorflow.keras import datasets
import tensorflow
from matplotlib import pyplot as plt
from sklearn.metrics import f1_score

# Load cifar 10 dataset
(x_train, y_train), (x_test, y_test) = datasets.cifar10.load_data()

# Normalize
x_train = x_train/255.0
x_test = x_test/255.0

# Perform one-hot encoding
y_train = utils.to_categorical(y_train)
y_test = utils.to_categorical(y_test)

print(y_train.shape)
```

```python
# This is a very very basic neural network classifier for demonstration purposes

def classifier():
    _input = layers.Input(shape=(x_train.shape[1],x_train.shape[2],x_train.shape[3]))
    
    conv=layers.Conv2D(32, (3,3), activation="relu") (_input)
    conv=layers.MaxPooling2D() (conv)
    
    conv=layers.Conv2D(64, (3,3), activation="relu") (conv)
    conv=layers.MaxPooling2D() (conv)
    
    conv=layers.Conv2D(128, (3,3), activation="relu") (conv)
    conv=layers.GlobalAveragePooling2D() (conv)
    
    dense = layers.Dense(50, activation="relu") (conv)
    dense = layers.Dense(10, activation="softmax") (dense)
    
    m = models.Model(_input, dense)
    
    m.compile(loss="categorical_crossentropy", optimizer="adam", metrics=["accuracy", "AUC", "Precision"])
    
    return m
```

```python
model = classifier()
print(model.summary())
```

```python
# Train the model
model.fit(x_train, y_train, batch_size=16, epochs=15)
```

```python
loss_object = tensorflow.keras.losses.CategoricalCrossentropy()

# This function will calculate the gradient signs for the attack
def create_adversarial_pattern(input_image, input_label):
    input_image = tensorflow.convert_to_tensor(input_image) # convert numpy array to tensor
    input_label = tensorflow.convert_to_tensor(input_label) # convert numpy array to tensor
    
    with tensorflow.GradientTape() as tape: # the gradient tape functions records operations for calculating the gradient
        tape.watch(input_image) # set "trace" to the input -> this function makes sure that the input to the network is recoreded
        prediction = model(input_image) # get the prediction
        loss = loss_object(input_label, prediction) # calculate the loss with the truth (input_label) and prediction
        
    # Get the gradients of the loss w.r.t to the input image.
    gradient = tape.gradient(loss, input_image)
    # Get the sign of the gradients to create the perturbation
    signed_grad = tensorflow.sign(gradient)
    
    return signed_grad

# Code from: Tensorflow (2020)
```

```python
image = x_test[0]
image = image[None,] # add batch dimension
label = y_test[0]
label = label[None,] # add batch dimension

# get the noise
perturbations = create_adversarial_pattern(image, label)
plt.imshow(perturbations[0]) # show nois
```

```python
# This function will display the image and the pertubated image

def display_images(epsilons, image, label):
    print("The true label is: " + str(np.argmax(label)))
    
    prediction = model.predict(image)
    print("The prediction for the original image is: " + str(np.argmax(prediction)) + " with " + str((prediction[0][np.argmax(prediction)])*100) + "% confidence" )
    
    plt.imshow(image[0])
    plt.title("Original image")
    plt.show()
    
    for i in range(0, len(epsilons)):
        eps=epsilons[i]
        perturbations = create_adversarial_pattern(image, label)
        adv_x = image + eps*perturbations
        prediction=model.predict(adv_x)
        print("The prediction is: " + str(np.argmax(prediction)) + " with " + str((prediction[0][np.argmax(prediction)])*100) + "% confidence, epsilon " + str(eps))
        plt.imshow(adv_x[0])
        plt.title("Pertubated images with epsilon " + str(eps))
        plt.show()
```

```python
img = x_test[25]
img = img[None, ]

l = y_test[25]
l=l[None, ]

eps_list=[0.001, 0.01, 0.05] # provide different strenght of the pertubation

display_images(epsilons=eps_list, image=img, label=l)
```
